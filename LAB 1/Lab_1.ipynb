## Understanding the Perceptron Model

This notebook demonstrates a simple implementation of a Perceptron, a fundamental building block in neural networks, often used for binary classification tasks. We will break down the code step-by-step, explaining each component.

### 1. Importing Libraries

We start by importing the necessary libraries. In this case, we only need `numpy` for numerical operations, especially for handling arrays and dot products.

```python
import numpy as np
```

### 2. The Step Activation Function

The Perceptron uses a simple step activation function. This function determines the output of the Perceptron based on the weighted sum of its inputs and a bias. If the sum is greater than or equal to zero, it outputs 1; otherwise, it outputs 0.

```python
def _step_function(x):
    """
    Step activation function.
    Returns 1 if x >= 0, otherwise returns 0.
    """
    return 1 if x >= 0 else 0
```

### 3. The Perceptron Class

This class encapsulates the Perceptron's logic, including its initialization, prediction mechanism, and training algorithm.

#### Initialization (`__init__`)

The constructor sets up the Perceptron's initial weights and bias. Weights are usually initialized randomly, but for simplicity, they are initialized to zeros here. The learning rate (`lr`) controls how much the weights and bias are adjusted during training.

#### Prediction (`predict`)

Given an input `x`, the `predict` method calculates the weighted sum of inputs plus the bias, and then applies the step activation function to produce a binary output (0 or 1).

#### Training (`train`)

The `train` method iterates through the training data for a specified number of `epochs`. In each iteration, it makes a prediction, calculates the error (difference between target and prediction), and updates the weights and bias if the prediction was incorrect. This process continues until the model converges (no errors in an epoch) or the maximum number of epochs is reached.

```python
class Perceptron:
    """
    A simple Perceptron model for binary classification.
    """

    def __init__(self, input_size, learning_rate=0.1):
        """
        Initializes the Perceptron with random weights and bias.

        Args:
            input_size (int): The number of features in the input data.
            learning_rate (float): The learning rate for weight updates.
        """
        # Initialize weights randomly to break symmetry, typically small values
        # For simplicity, starting with zeros as in the original code, but random is often better.
        self.weights = np.zeros(input_size)
        # Initialize bias to 0
        self.bias = 0
        # Set the learning rate
        self.lr = learning_rate
        print(f"Perceptron initialized with input_size={input_size}, learning_rate={self.lr}")

    def predict(self, x):
        """
        Makes a prediction for a single input vector x.

        Args:
            x (np.array): A single input feature vector.

        Returns:
            int: The predicted class (0 or 1).
        """
        # Calculate the weighted sum of inputs and add bias
        # z = w1*x1 + w2*x2 + ... + wn*xn + bias
        z = np.dot(x, self.weights) + self.bias
        # Apply the step activation function
        return _step_function(z)

    def train(self, X, y, epochs=10):
        """
        Trains the Perceptron model using the given training data.

        Args:
            X (np.array): Training input features (N_samples, N_features).
            y (np.array): Target labels (N_samples,).
            epochs (int): The number of training iterations over the entire dataset.
        """
        print(f"\nStarting training for {epochs} epochs...")
        for epoch in range(epochs):
            total_error = 0
            for i, (xi, target) in enumerate(zip(X, y)):
                # Make a prediction with current weights
                prediction = self.predict(xi)
                # Calculate the error
                error = target - prediction
                total_error += abs(error)

                # Update weights and bias based on the error
                # If prediction is wrong, adjust weights and bias
                if error != 0:
                    # Update weights: weights = weights + learning_rate * error * input
                    self.weights += self.lr * error * xi
                    # Update bias: bias = bias + learning_rate * error
                    self.bias += self.lr * error
            print(f"Epoch {epoch + 1}/{epochs}, Total Error: {total_error}, Weights: {self.weights}, Bias: {self.bias:.2f}")
            # If total_error is 0, it means all samples are correctly classified
            if total_error == 0:
                print("Converged! No errors in this epoch.")
                break
        print("Training complete.")
```

### 4. Training the Perceptron (AND Gate Example)

Now, let's put the Perceptron to use. We'll train it to implement the logical AND gate. The AND gate outputs 1 only if both inputs are 1; otherwise, it outputs 0.

#### Define Training Data

`X_train` contains the input combinations (e.g., [0, 0], [0, 1]) and `y_train` contains the corresponding target outputs for the AND gate.

```python
# --- 1. Define Training Data (AND gate example) ---
# Input features (X) for the AND gate
X_train = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
# Target labels (y) for the AND gate
y_train = np.array([0, 0, 0, 1])

print("--- Perceptron for AND Gate ---")
print("Training Data (X):\n", X_train)
print("Target Labels (y):", y_train)
```

#### Initialize and Train

We create an instance of our `Perceptron` class with two input features and a learning rate, then train it using our `X_train` and `y_train` data.

```python
# --- 2. Initialize and Train the Perceptron ---
# Create a Perceptron instance with 2 input features
perceptron = Perceptron(input_size=2, learning_rate=0.1)

# Train the perceptron for a specified number of epochs
perceptron.train(X_train, y_train, epochs=10)
```

### 5. Final Model Parameters

After training, we can inspect the final weights and bias that the Perceptron learned. These parameters define the decision boundary that separates the classes.

```python
print("\n--- Final Model Parameters ---")
print(f"Final Weights: {perceptron.weights}")
print(f"Final Bias: {perceptron.bias:.2f}")
```

### 6. Test Predictions and Accuracy

Finally, we test the trained Perceptron on the same training data to see how well it learned the AND gate logic. We also calculate the accuracy.

```python
# --- 3. Test Predictions ---
print("\n--- Predictions on Training Data ---")
correct_predictions = 0
for xi, target in zip(X_train, y_train):
    prediction = perceptron.predict(xi)
    print(f"Input: {xi}, Target: {target}, Predicted: {prediction}")
    if prediction == target:
        correct_predictions += 1

# --- 4. Calculate and Display Accuracy ---
accuracy = (correct_predictions / len(y_train)) * 100
print(f"\nAccuracy on training data: {accuracy:.2f}%")

# You can also test with new data if needed
# print("\n--- Testing with a new input (e.g., [0,1]) ---")
# new_input = np.array([0, 1])
# print(f"Input: {new_input}, Predicted: {perceptron.predict(new_input)}")
```
