## Neural Network Implementations (PyTorch and TensorFlow/Keras)

This notebook demonstrates the implementation of a simple feedforward neural network using both PyTorch and TensorFlow/Keras, specifically to solve an AND logic gate problem.

```python
### Implement a feedforward neural network using TensorFlow or PyTorch.
# PyTorch Implementation
import torch
import torch.nn as nn
import torch.optim as optim

# Sample dataset: AND logic gate
X = torch.tensor([[0,0], [0,1], [1,0], [1,1]], dtype=torch.float32)
y = torch.tensor([[0], [0], [0], [1]], dtype=torch.float32)

# Define a feedforward neural network
class FeedforwardNN(nn.Module):
    def __init__(self):
        super(FeedforwardNN, self).__init__()
        self.fc1 = nn.Linear(2, 8) # Increased hidden units from 4 to 8
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(8, 1) # Adjusted input for second layer to match
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.relu(self.fc1(x))
        out = self.sigmoid(self.fc2(out))
        return out

# Initialize model, loss, and optimizer
model = FeedforwardNN()
criterion = nn.BCELoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)

# Train the model
for epoch in range(200): # Increased epochs from 100 to 200
    outputs = model(X)
    loss = criterion(outputs, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

# Predict
print("\nPredictions:")
with torch.no_grad():
    predictions = model(X)
    for i, pred in enumerate(predictions):
        print(f"Input: {X[i].tolist()} => Predicted: {round(float(pred), 4)} => Class: {int(pred >= 0.5)}")
```

### PyTorch Model for AND Gate

This section implements a feedforward neural network using PyTorch. It defines a simple `FeedforwardNN` class, trains it on an AND gate dataset using Binary Cross-Entropy Loss and the Adam optimizer, and then prints the predictions for the input combinations.

### PyTorch Implementation Results

The PyTorch model successfully learned the AND logic. As observed, inputs `[0,0]`, `[0,1]`, and `[1,0]` result in predictions close to 0 (classified as 0), while the input `[1,1]` results in a prediction close to 1 (classified as 1).

### TensorFlow/Keras Multilayer Perceptron

This section demonstrates how to implement a Multilayer Perceptron (MLP) using TensorFlow's Keras API to solve the same AND logic gate problem. It constructs a sequential model with two dense hidden layers and an output layer, compiles it with Adam optimizer and binary cross-entropy loss, and then trains and evaluates the model.

```python
# Multilayer Perceptron with Keras (TensorFlow)
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Example dataset: AND gate
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=np.float32)
y = np.array([[0], [0], [0], [1]], dtype=np.float32)

# Build MLP model
model = Sequential([
    Dense(8, input_dim=2, activation='relu'),  # Hidden Layer 1
    Dense(4, activation='relu'),               # Hidden Layer 2
    Dense(1, activation='sigmoid')             # Output Layer
])


# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
model.fit(X, y, epochs=100, verbose=1)

# Evaluate and predict
print("\nPredictions:")
predictions = model.predict(X)
for i in range(len(X)):
    print(f"Input: {X[i]} => Predicted: {predictions[i][0]:.4f} => Class: {int(predictions[i][0] >= 0.5)}")
```

### TensorFlow/Keras Implementation Results

The TensorFlow/Keras Multilayer Perceptron also successfully learned the AND logic. Similar to the PyTorch model, it correctly classifies `[0,0]`, `[0,1]`, and `[1,0]` as 0, and `[1,1]` as 1, demonstrating its ability to learn the non-linear relationship.
